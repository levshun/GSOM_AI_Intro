## 1.2 Where does data come from?

Every data project starts with one of three sources:

1. **Existing files** - a messy CSV, an export from a system, a spreadsheet.
   Most real-life data is like this: full of surprises and mistakes.
2. **Synthetic data** - data a program or an AI *creates* so you can test ideas
   without private or expensive data, and with all the edge cases you want.
3. **Web data** - data scraped from web pages. Powerful but legally and
   technically tricky (we explore it conceptually in Part 3).

Regardless of the source, your goal is usually the same: **tidy data**.
In tidy data every column is one variable, every row is one observation, and
nothing is duplicated. In Part 3 you will turn a messy table into a tidy one.

## 1.4 Anatomy of a good prompt

A prompt is instructions to an AI. For data work the **CATA** recipe works well:

- **C**ontext - the situation: which project, which file, what you are trying to do.
- **A**ction - one clear task, with the exact output format/path asked for.
- **T**one - "short", "beginner-friendly", "keep the code commented".
- **A**nticipate - edge cases: "keep it offline", "flag outliers instead of deleting".

Plus one extra piece that makes this course *results-driven*: a **checkpoint**.
Tell the assistant what you will verify afterwards ("tell me the path, row count,
and 3 messy things"). A good prompt is testable - you can check whether it worked.

> **Prompting is the small part.** The real skill is *analyzing the result*:
> did the assistant produce a valid file? Are the claims in its answer true in
> the file? Each checkpoint cell in this course is exactly that habit.

## 1.5 Part 1 - key takeaways

- Data comes from **existing files**, **synthetic data**, or **web data**.
- **Tidy data**: columns = variables, rows = observations, no duplicates.
- `AGENTS.md` gives project rules; `.opencode/agent/*.md` defines specialists.
- Agent changes need an **OpenCode restart**.
- A prompt = CATA + a **checkpoint** you can verify.
- Always **check the artifact**, never trust the assistant's summary alone.

## 2.2 Why synthetic data, and how to design it

Synthetic data is *fake but realistic* data, created on purpose. We use it to:

- avoid **privacy** problems (never use real names, emails, or PII),
- test edge cases (missing values, outliers, duplicates) cheaply,
- get as much data as we want without a real source,
- share examples freely.

The rule before writing any generator (or asking an AI): **decide the schema
first**. A schema says which columns exist and their constraints:

| Column | Type | Constraint |
|--------|------|-----------|
| customer_id | text | unique, `C0001...` |
| name | text | two words, no real people |
| email | text | looks like an email |
| country | text | a small fixed list |
| age | int | 18-90 |
| signup_date | date | recent ISO dates |
| spend | float | positive, 2 decimals |

Once the schema is fixed, values are just random draws inside those rules.
Setting a fixed `random.seed(...)` makes the whole thing **reproducible** - anyone
running it gets the same "random" numbers.

Our generator below builds the customers table exactly this way.

```python
import random
import datetime as dt
import pandas as pd

FIRST = ["Anna", "Ben", "Cara", "Denis", "Eva", "Finn", "Gia", "Hugo", "Iris", "Jon"]
LAST  = ["Meyer", "Novak", "Osei", "Paoli", "Quinn", "Reyes", "Singh", "Tate", "Ueda", "Voss"]
COUNTRIES = ["US", "CA", "UK", "MX", "DE", "FR"]

def make_customers(n: int, seed: int) -> pd.DataFrame:
    """Generate n synthetic customers with a fixed seed (reproducible)."""
    rng = random.Random(seed)
    rows = []
    for i in range(n):
        rows.append({
            "customer_id": f"C{i + 1:04d}",
            "name": f"{rng.choice(FIRST)} {rng.choice(LAST)}",
            "email": f"C{i + 1:04d}@example.com",   # safe placeholder, never a real address
            "country": rng.choice(COUNTRIES),
            "age": rng.randint(18, 90),
            "signup_date": dt.date(2023, 1, 1) + dt.timedelta(days=rng.randrange(900)),
            "spend": round(rng.uniform(0, 500), 2),
        })
    return pd.DataFrame(rows)

# Write a deterministic synthetic dataset (200 rows) and keep it reproducible.
df_synth = make_customers(200, seed=42)
df_synth.to_csv("data/generated/customers_synth.csv", index=False)
print("Wrote data/generated/customers_synth.csv", df_synth.shape)
df_synth.head(5)
```

```python
# Quick "results analysis" of our own generated file - look before you trust.
df_synth = pd.read_csv("data/generated/customers_synth.csv")

print("Shape:", df_synth.shape)
print("\nNulls per column:\n", df_synth.isna().sum().to_dict())
print("\nDuplicates:", int(df_synth.duplicated().sum()))
print("\nData types:\n", df_synth.dtypes.to_string())
print("\nNumeric ranges:\n", df_synth[["age", "spend"]].describe().T[["min", "max", "mean"]])
print("\nCountries:", sorted(df_synth["country"].unique()))
```

### Why we check before trusting

Notice what the analysis cell did: it *did not* assume the file is good. It
asked the data questions itself - shape, nulls, duplicates, types, ranges,
countries. Two things worth remembering:

- A generator can silently produce zeros, negative values, or duplicates. Checking is cheap; trusting is expensive.
- The schema check above is exactly the same check you will run on the AI's output next.

```python
# Compare the two synthetic files side by side (results analysis).
import pandas as pd

mine = pd.read_csv("data/generated/customers_synth.csv")
ai = df_ai

def profile(df: pd.DataFrame) -> dict:
    return {
        "rows": df.shape[0],
        "cols": df.shape[1],
        "nulls": int(df.isna().sum().sum()),
        "duplicates": int(df.duplicated().sum()),
    }

compare = pd.DataFrame({"deterministic script": profile(mine),
                        "AI-assisted": profile(ai)}).T
print(compare.to_string())
print("\nShared columns:", sorted(set(mine.columns) & set(ai.columns)))
print("Only in mine:", sorted(set(mine.columns) - set(ai.columns)) or "none")
print("Only in AI  :", sorted(set(ai.columns) - set(mine.columns)) or "none")
```

```python
# Only meaningful with a REAL AI-generated file. The fallback run is skipped so
# students are not shown two identical charts.
if used_fallback:
    print("Skipped side-by-side chart: the AI file was missing during this run.")
    print("Re-run these cells after the AI task to see a real comparison.")
else:
    import matplotlib.pyplot as plt
    import numpy as np

    cols = [c for c in ["email", "age", "spend", "country"]
            if c in mine.columns and c in ai.columns]
    m = mine[cols].isna().sum()
    a = ai[cols].isna().sum()

    fig, ax = plt.subplots(figsize=(7.5, 3.8))
    x = np.arange(len(cols)); w = 0.35
    ax.bar(x - w / 2, m, w, label="deterministic script", color="#4c72b0")
    ax.bar(x + w / 2, a, w, label="AI-assisted", color="#dd8452")
    ax.set_xticks(x); ax.set_xticklabels(cols)
    ax.set_ylabel("missing values")
    ax.set_title("Missing values per column: script vs AI")
    ax.legend(); plt.tight_layout(); plt.show()
```

### Reading the comparison

Even when both files "look fine", the comparison can differ in real ways:

- **Schema drift** - the AI might add or rename a column or use a different date format. That breaks your pipeline later, so note it now.
- **Ranges** - maybe the AI lets `age` reach 18-80 while yours is 18-90. Not wrong, but a *difference* you must know about.
- **Edge cases** - a good generator includes a couple of missing values; too many are a red flag, none at all is suspicious.
- **Privacy** - verify that no real-looking personal data slipped in.

This is what **results analysis** means for generated data: don't just eyeball the CSV, *measure* the schema, the shape, and the constraints - and make the AI tell you its assumptions so you can check them.

## 2.4 Part 2 - key takeaways

- Synthetic data protects privacy, creates edge cases, and is reproducible with a `seed`.
- Always **design the schema** (columns, types, ranges) before generating.
- Generate once yourself, once with AI; then **compare** profile metrics (rows, cols, nulls, duplicates, ranges).
- The AI's answer includes *claims*; your checkpoints confirm or deny them.

## 3.2 How web data collection works (conceptually)

When you look at a website, your browser asks a server for a page using **HTTP**
and receives **HTML** - the markup that structures the page (headings, tables,
links). Instead of reading it with your eyes, a program can parse that HTML and
extract the table cells. That is **scraping**.

Scraping is powerful but has firm rules:

- **robots.txt** - a file a site publishes saying what automatic access is allowed.
- **Terms of service** - even if a page loads, you must follow the site's rules.
- **Rate limits & load** - hammering a server with requests is rude and can be illegal.
- **Copyright & personal data** - collected data may be subject to rights.

That is why this course does **not** fetch live websites. We use a **local HTML
snapshot** (`data/sample_market.html`) saved beforehand. The skill you practise
here - parse, structure, clean - is the same skill you would use on a real page
you are allowed to collect.

The cell below "scrapes" that local file: it reads the HTML, finds the table, and
turns it into a pandas DataFrame.

```python
from bs4 import BeautifulSoup
import pandas as pd

# Read the LOCAL snapshot (we never hit the live web in this course).
with open("data/sample_market.html", encoding="utf-8") as fh:
    soup = BeautifulSoup(fh.read(), "html.parser")

table = soup.find("table", id="prices")
headers = [th.get_text(strip=True) for th in table.find("thead").find_all("th")]

rows = []
for tr in table.find("tbody").find_all("tr"):
    cells = [td.get_text(strip=True) for td in tr.find_all("td")]
    rows.append(cells)

prices = pd.DataFrame(rows, columns=headers)
print(f"Parsed {len(prices)} rows from {len(soup.find_all('table'))} table(s):\n")
print(prices.to_string(index=False))

# A tiny "cleaning" of what we scraped: prices still have commas.
prices["Price (USD)"] = pd.to_numeric(prices["Price (USD)"].str.replace(",", ""), errors="coerce")
prices["In stock"] = pd.to_numeric(prices["In stock"], errors="coerce")
print("\nAfter converting text prices to numbers:\n")
print(prices.dtypes.to_string())
```

```python
import pandas as pd
import numpy as np
import re
import warnings

RAW = "data/messy_customers.csv"
# The ASSISTANT's task (Your turn) writes data/cleaned/customers_clean.csv.
# This fallback run writes its own clearly-labelled copy.
OUT = "data/cleaned/customers_clean_manual.csv"

raw = pd.read_csv(RAW)
report = {"start_rows": raw.shape[0]}   # we record every decision

# --- Step 1: normalize text (names, emails, country) -----------------------
clean = raw.copy()
clean["name"] = clean["name"].str.strip().str.replace(r"\s+", " ", regex=True).str.title()
clean["email"] = clean["email"].str.strip().str.lower()

# A missing email is any blank value or anything without an @ -> NaN (one meaning).
clean.loc[~clean["email"].str.contains("@", na=True), "email"] = np.nan

def norm_country(x):
    """Collapse country variants to one spelling; keep missing values missing."""
    if pd.isna(x):
        return np.nan
    mapping = {"US": "US", "USA": "US", "U.S.A.": "US", "usa": "US", "us": "US",
               "United States": "US",
               "CA": "CA", "Canada": "CA", "ca": "CA",
               "MX": "MX", "Mexico": "MX",
               "UK": "UK"}
    return mapping.get(str(x).strip(), str(x).strip())
clean["country"] = clean["country"].map(norm_country)

# --- Step 2: parse spend into numbers and flag outliers ---------------------
def parse_spend(s):
    """Turn a messy spend cell into a (number, flag) pair."""
    if pd.isna(s):
        return pd.NA, ""
    if isinstance(s, (int, float)):
        v = float(s)
    else:
        t = str(s).strip().upper()
        if t in ("", "N/A", "NA", "UNKNOWN", "NULL", "-"):
            return pd.NA, ""
        t = t.replace("$", "").replace("USD", "").replace(" ", "")
        # commas are thousands separators only when they group three digits,
        # e.g. 2,000,000 -> 2000000; a lone decimal comma (5,00) stays a dot.
        if "," in t:
            if re.match(r"^\d+(,\d{3})+(\.\d+)?$", t):
                t = t.replace(",", "")
            else:
                t = t.replace(",", ".")
        try:
            v = float(t)
        except ValueError:
            return pd.NA, "unparsed"
    flag = "negative" if v < 0 else ("huge" if v > 100_000 else "ok")
    return v, flag

parsed = [parse_spend(x) for x in clean["spend"]]
clean["spend"] = [p[0] for p in parsed]
clean["spend_flag"] = [p[1] for p in parsed]

# --- Step 3: parse dates into real datetime (NA for unparseable) ------------
def parse_date(s):
    """Parse a messy date to datetime, or mark it 'unparsed' rather than guessing."""
    if pd.isna(s):
        return pd.NaT, ""
    t = str(s).strip()
    if t.lower() in ("unknown", "n/a", "na", "null", ""):
        return pd.NaT, ""
    # Suppress pandas's day-first guessing notice; ambiguous dates become NaN.
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", UserWarning)
        try:
            return pd.to_datetime(t), ""
        except Exception:
            return pd.NaT, "unparsed"
dates = [parse_date(x) for x in clean["signup_date"]]
clean["signup_date"] = pd.to_datetime([d[0] for d in dates])
clean["date_flag"] = [d[1] for d in dates]

report = {**report,
          "emails_missing": int(clean["email"].isna().sum()),
          "unparsed_dates": int((clean["date_flag"] == "unparsed").sum()),
          "unparsed_spend": int((clean["spend_flag"] == "unparsed").sum()),
          "negative_spend": int((clean["spend_flag"] == "negative").sum()),
          "huge_spend": int((clean["spend_flag"] == "huge").sum()),
          }

# --- Step 4: drop duplicate emails (keep first), but KEEP missing-email rows --
# Note: pandas treats NaN as a "duplicate" too, so we guard with notna() to
# avoid deleting customers whose email we simply do not know.
key = clean["email"]
is_dup = key.notna() & key.duplicated(keep="first")
report["duplicate_rows_removed"] = int(is_dup.sum())
clean = clean[~is_dup].reset_index(drop=True)

print("Cleaning steps done. Report so far:\n")
for k, v in report.items():
    print(f"  {k:22s} {v}")
```

```python
import os

# Save the manual cleaned table + metrics + a short human report.
os.makedirs("data/cleaned", exist_ok=True)
clean[["customer_id", "name", "email", "signup_date", "spend", "spend_flag",
       "country", "region"]].to_csv(OUT, index=False)

report["end_rows"] = clean.shape[0]
report["missing_after"] = int(clean.isna().sum().sum())

report_path = "data/cleaned/customers_clean_manual_report.md"
with open(report_path, "w", encoding="utf-8") as fh:
    fh.write("# Cleaning report (manual fallback)\n\n")
    fh.write("- Rows: start %d -> end %d\n" % (report["start_rows"], report["end_rows"]))
    fh.write("- Missing values total: before %d -> after %d\n" % (raw.isna().sum().sum(), report["missing_after"]))
    fh.write("- Emails missing: %d\n" % report["emails_missing"])
    fh.write("- Unparsed dates: %d, unparsed spend: %d\n" % (report["unparsed_dates"], report["unparsed_spend"]))
    fh.write("- Negative spend: %d, huge spend flagged: %d\n" % (report["negative_spend"], report["huge_spend"]))

print("Saved:", OUT)
print("Saved:", report_path, "\n")
print("Manual result preview:\n")
clean.head(8)
```

### Explaining each cleaning step

1. **Normalize text** - trimming spaces and fixing casing makes one record unique
   instead of "John Doe" vs "john doe" being two records.
2. **Parse numbers** - spend was text with `$`, commas and units; we turned it
   back into numbers, and kept a `spend_flag` for values we refused to accept as
   clean (negatives, huge outliers, unparsed).
3. **Parse dates** - mixed date strings became one `datetime` type; anything we
   could not parse was marked, not guessed.
4. **Remove duplicates** - after normalizing emails, the same person can appear
   several times; we keep the first row and record how many were dropped.
5. **Flag, don't delete** - outliers stay in the data but carry a flag so a later
   step (e.g. the pipeline in Part 4) can decide what to do with them. Deleting
   silently loses information - and makes mistakes invisible.

Now let's visualize the improvement and verify the numbers.

### Cheat sheet: the pandas moves we just used

| Goal | One-liner |
|------|-----------|
| Read a file | `df = pd.read_csv("path.csv")` |
| Look first | `df.head()`, `df.info()`, `df.describe()` |
| Missing counts | `df.isna().sum()` |
| Strip whitespace | `df["col"].str.strip()` |
| Fix casing | `df["col"].str.title()` (or `.str.lower()`) |
| Unique count | `df["col"].nunique()` |

| Parse dates | `pd.to_datetime(df["col"], errors="coerce")` |
| Numbers from text | `pd.to_numeric(df["col"].str.replace("$", ""), errors="coerce")` |
| Flag, not delete | `df["flag"] = df["col"] > threshold` |
| Save | `df.to_csv("out.csv", index=False)` |

## 4.2 The pipeline: generate -> clean -> validate -> save

Real work is not done cell by cell. You package the steps we already ran into
functions and call them in a fixed order. If any step fails, you know the data
should not be trusted.

<a id="part4"></a>

```python
# Reuse the functions already defined: make_customers (Part 2) and the cleaning
# steps (Part 3). We package the whole thing as one repeatable pipeline.

def validate(df: pd.DataFrame, id_col: str = "customer_id") -> dict:
    """Return simple, numeric quality metrics for a dataset."""
    n = df.shape[0]
    missing = int(df.isna().sum().sum())
    completeness = round(1 - missing / (n * df.shape[1]), 4)
    duplicates = int(df.duplicated().sum())
    uniq_ids = int(df[id_col].nunique())
    return {
        "rows": n,
        "missing_cells": missing,
        "completeness": completeness,
        "duplicate_rows": duplicates,
        "unique_ids": uniq_ids,
        "unique_id_ratio": round(uniq_ids / max(1, n), 4),
    }

def pipeline(n_rows: int = 200, seed: int = 42) -> dict:
    """Full run: generate fresh data, clean it, validate it, save outputs."""
    # 1) generate
    df = make_customers(n_rows, seed=seed)
    # 2) clean (the lightweight version from Part 3, for generated data)
    df = df.drop_duplicates().reset_index(drop=True)
    df["name"] = df["name"].str.strip().str.title()
    # 3) validate
    metrics = validate(df)
    # 4) save
    df.to_csv("data/cleaned/pipeline_customers.csv", index=False)
    return {**metrics, "seed": seed}

print("Running the reproducible pipeline...")
result = pipeline()
for k, v in result.items():
    print(f"  {k:18s} {v}")
print("\nSaved -> data/cleaned/pipeline_customers.csv")
```

```python
# End-to-end: validate the MESSY file we cleaned by hand in Part 3 too.
import os

p = "data/cleaned/customers_clean_manual.csv"
if os.path.exists(p):
    clean_df = pd.read_csv(p)
    m = validate(clean_df)
    print("Cleaned messy data - quality metrics (validate() reused over Part 3 output):")
    for k, v in m.items():
        print(f"  {k:18s} {v}")
else:
    print("customers_clean_manual.csv not found. Run the Part 3 cells first.")
```

### What the validation numbers mean

- **Completeness** - 1.0 means zero missing cells; 0.95 means 5% of cells are empty.
- **Uniqueness** - `unique_id_ratio == 1.0` means every record has a different ID.
  Under 1.0 is a serious problem (duplicates broke the unique key).
- **Duplicate rows** - exact repeated rows; on tidy data this should be 0.
- **Sensible ranges** - spot-check `age` and `spend` (not in the table above,
  but the reviewer agent will check them).

A pipeline is trustworthy when these numbers stay within your chosen thresholds
- and a reviewer *independently* confirms them.

### Reading a report critically (even an AI's)

A good review report is a list of *claims with numbers*. Treat it like this:

1. **Reproduce it** - the fastest way to trust a report is to compute the same
   number yourself (the checkpoint cells do that).
2. **Find the disagreements** - if the report says "no missing values" but your
   number shows 5, the report is wrong (or looked at a different file).
3. **Decide what passes** - set your own thresholds: "completeness >= 0.98 and
   unique_id_ratio == 1.0". Anything below is a reason to clean more.
4. **Remember the limits** - even a verified dataset has *unknown unknowns*:
   bias baked into the source, wrong units, or a schema change later on.

## 4.5 Ethics and limitations to remember

- **Scraping is regulated.** robots.txt, terms of service, rate limits and
  copyright apply even if a page "loads in your browser". When in doubt,
  use official APIs or ask permission.
- **LLMs can fabricate data.** An AI may invent a *plausible-looking* schema,
  numbers, or even citations. Always validate generated data against your own
  checks and a known-good example.
- **PII is sensitive.** Real names, emails, addresses must not appear in
  synthetic data or in shared outputs.
- **Prompt length costs tokens.** Large context and long prompts increase cost
  and latency; keep prompts focused (the checkpoint idea helps here too).
- **Review the agent's work.** An agent that can edit files must have its
  changes checked: cleaned data, saved reports, and especially deleted rows.
